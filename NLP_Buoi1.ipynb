## 5.Tinh toan do thi

```python
# Create tensors with requires_grad=True to track operations
x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(3.0, requires_grad=True)
print(f"x = {x}")
print(f"y = {y}")
# Build a computational graph
z = x**2 + y**3
print(f"z = x^2 + y^3 = {z}")
# Compute gradients
z.backward()
# Access gradients
print(f"Gradient of z with respect to x (dz/dx): {x.grad}")
print(f"Gradient of z with respect to y (dz/dy): {y.grad}")
```

```python
# Gradient accumulation
print("Gradient accumulation:")
# Reset gradients
x.grad.zero_()
y.grad.zero_()
print(f"After zeroing gradients:")
print(f"x.grad = {x.grad}")
print(f"y.grad = {y.grad}")
# Compute gradients multiple times
z = x**2 + y**3
z.backward()
print(f"\nAfter first backward pass:")
print(f"x.grad = {x.grad}")
z = x**2 + y**3
z.backward()
print(f"\nAfter second backward pass (gradients are accumulated):")
print(f"x.grad = {x.grad}")
```

```python
# Detach a tensor from the graph
a = x.detach()
print(f"Detached tensor a = {a}")
print(f"a.requires_grad = {a.requires_grad}")
```

## 6.Vi du truc quan

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

# Create a range of x values
x_range = torch.linspace(-3, 3, 100, requires_grad=True)

# Define a function: f(x) = x^2
y = x_range**2

# Compute gradients for each x value
gradients = []
for i in range(len(x_range)):
    if x_range.grad is not None:
        x_range.grad.zero_()

    # Tính toán y_i và backward bên trong vòng lặp
    y_i = x_range[i]**2
    y_i.backward(retain_graph=True)
    gradients.append(x_range.grad[i].item())

# Convert to NumPy for plotting
x_np = x_range.detach().numpy()
y_np = y.detach().numpy()
gradients_np = np.array(gradients)

# Plot the function and its gradient
plt.figure(figsize=(10, 6))
plt.plot(x_np, y_np, 'b-', label='f(x) = x^2')
plt.plot(x_np, gradients_np, 'r-', label="f'(x) = 2x")
plt.grid(True)
plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
plt.axvline(x=0, color='k', linestyle='-', alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Function and its Gradient')
plt.legend()
plt.show()
```

```python
a = torch.tensor([1.0, 2.0], requires_grad=True)
b = torch.tensor([3.0, 4.0])

try:
    # Thử cộng in-place vào một biến đang track gradient
    a.add_(b) 
except RuntimeError as e:
    print(f"Lỗi xuất hiện: {e}")

# Cách đúng: Tạo ra tensor mới c
c = a + b
print("Cộng bình thường (không in-place) thì OK!")
```

Khi nào NÊN dùng In-place?
Lý do duy nhất và quan trọng nhất để dùng in-place là Tiết kiệm bộ nhớ.   
Dữ liệu cực lớn: Khi bạn đang làm việc với các Tensor khổng lồ (như ảnh độ phân giải cao) trên chiếc Acer Nitro 5, bộ nhớ VRAM của GPU có hạn. Dùng in-place giúp bạn không phải tạo ra các bản sao trung gian, tránh lỗi "Out of Memory" (OOM).   
Các tầng không cần tính Gradient: Ví dụ trong các bước tiền xử lý dữ liệu hoặc ở cuối quá trình inference (dự đoán), nơi bạn không cần cập nhật trọng số.

2. Giải thích cơ chế tích lũy Gradient của PyTorch
Tại sao bị cộng dồn thay vì ghi đè?
Trong PyTorch, khi bạn gọi .backward(), giá trị gradient mới tính được sẽ được cộng thêm vào thuộc tính .grad hiện có của Tensor (toán tử +=).  
Lý do thiết kế này tồn tại:   
Tiết kiệm bộ nhớ (Memory efficiency): Khi bạn có một Batch dữ liệu cực lớn không thể nhét vừa VRAM của Nitro 5, bạn có thể chia nhỏ Batch đó ra. Bạn tính gradient cho từng phần nhỏ, cộng dồn chúng lại, rồi mới cập nhật trọng số một lần. Đây gọi là kỹ thuật Gradient Accumulation.  
Hỗ trợ đồ thị phức tạp: Một Tensor có thể tham gia vào nhiều nhánh tính toán khác nhau. Việc cộng dồn giúp tổng hợp đầy đủ đạo hàm từ tất cả các nhánh đó về một mối.

```python
x = torch.tensor(2.0, requires_grad=True)
tong_grad = 0.0

for i in range(1, 4):
    # BƯỚC QUAN TRỌNG: Reset gradient về 0 trước khi tính lần mới
    if x.grad is not None:
        x.grad.zero_() 

    f = i * x**2
    f.backward()
    
    current_grad = x.grad.item()
    tong_grad += current_grad
    print(f"Lần {i}: x.grad = {current_grad:.1f}")

print(f"Tổng gradient (Mong muốn: 24.0): {tong_grad:.1f}")
# Kết quả bây giờ sẽ là: 4.0 + 8.0 + 12.0 = 24.0
```

```python
import torch
import matplotlib.pyplot as plt

# Khởi tạo w tại 0.0
w = torch.tensor(0.0, requires_grad=True)
lr = 0.1 # Learning rate
lich_su_loss = []
lich_su_w = []

for buoc in range(50):
    # 1. Tính L(w) theo công thức đề bài
    L = (w - 3)**2 + 2*(w - 3) + 1
    
    # 2. Xóa gradient cũ (tránh tích lũy cộng dồn như bài 11)
    if w.grad is not None:
        w.grad.zero_()
        
    # 3. Tính gradient (Backward pass)
    L.backward()
    
    # 4. Cập nhật w (Gradient Descent step)
    # Dùng no_grad để thực hiện phép trừ mà không tạo thêm đồ thị tính toán
    with torch.no_grad():
        w -= lr * w.grad 
        
    lich_su_loss.append(L.item())
    lich_su_w.append(w.item())

print(f"w tối ưu tìm được: {w.item():.4f} (kỳ vọng: 2.0)")

# --- Vẽ đồ thị kiểm chứng ---
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(lich_su_loss); axes[0].set_title("Loss theo bước")
axes[0].set_xlabel("Bước"); axes[0].set_ylabel("L(w)")

axes[1].plot(lich_su_w); axes[1].set_title("w theo bước")
axes[1].axhline(2.0, color="red", ls="--", label="w*=2")
axes[1].legend()
plt.tight_layout(); plt.show()
```

Loss: Thay vì giảm dần về 0, giá trị Loss sẽ tăng vọt lên những con số khổng lồ (ví dụ: $10^{10}, 10^{20}...$) và cuối cùng có thể hiện là inf hoặc NaN.Giá trị $w$: Thay vì tiến về $2.0$, $w$ sẽ nhảy qua nhảy lại quanh điểm tối ưu với khoảng cách ngày càng xa (ví dụ: đang ở $0$, nhảy sang $5$, rồi nhảy sang $-10$, rồi $30...$).

```python
w = torch.tensor(0.0, requires_grad=True)
lr = 1.5 # Learning rate quá lớn

for buoc in range(5): # Chỉ cần 5 bước là thấy thảm họa
    L = (w - 3)**2 + 2*(w - 3) + 1
    if w.grad is not None: w.grad.zero_()
    L.backward()
    
    with torch.no_grad():
        w -= lr * w.grad
        
    print(f"Bước {buoc+1}: w = {w.item():.2f}, Loss = {L.item():.2f}")
```

```python
# 1. Khởi tạo tensor z với các giá trị yêu cầu
# Lưu ý: cần requires_grad=True để tính được đạo hàm bậc một
z = torch.tensor([0.0, 1.0, 2.0, 5.0], requires_grad=True)

# 2. Định nghĩa hàm Sigmoid thủ công: sigma(z) = 1 / (1 + exp(-z))
# Chúng ta dùng torch.exp để PyTorch có thể theo dõi gradient
sigma_z = 1 / (1 + torch.exp(-z))

# 3. Tính đạo hàm bằng autograd
# Vì sigma_z là một vector, ta cần truyền một vector grad_outputs (thường là toàn số 1)
# Hoặc đơn giản là tính tổng trước khi backward để đưa về scalar
sigma_z.backward(torch.ones_like(z))

# 4. Hiển thị kết quả
print(f"{'z':>5} | {'Sigmoid sigma(z)':>18} | {'Đạo hàm sigma\'(z)':>18}")
print("-" * 50)
for i in range(len(z)):
    print(f"{z[i].item():5.1f} | {sigma_z[i].item():18.4f} | {z.grad[i].item():18.4f}")
```

```python
import torch

diem_kiem_tra = [0.0, 1.0, 2.0, 5.0]

print(f"{'z':>6} {'sigma(z)':>10} {'sigma_prime(z) [autograd]':>26} {'sigma_prime(z) [tay]':>22}")

for z_val in diem_kiem_tra:
    # Khởi tạo tensor cho từng giá trị và bật tính năng tính gradient
    z = torch.tensor(z_val, requires_grad=True)
    
    # 1. Tính sigma(z) thủ công không dùng torch.sigmoid
    sigma = 1 / (1 + torch.exp(-z))
    
    # 2. Tính đạo hàm bằng autograd
    sigma.backward()
    grad_autograd = z.grad.item()
    
    # 3. Tính đạo hàm bằng công thức tay: sigma(z)*(1 - sigma(z))
    # Lưu ý: dùng .item() để lấy giá trị số thuần túy
    s = sigma.item()
    grad_tay = s * (1 - s)
    
    print(f"{z_val:>6.1f} {sigma.item():>10.6f} {grad_autograd:>26.6f} {grad_tay:>22.6f}")
```

```python
# ---- Dữ liệu tổng hợp (đừng thay đổi) ----
torch.manual_seed(42)
X = torch.linspace(-3, 3, 100) # shape: (100,)
y_true = 2.5 * X - 1.0         # w_that=2.5, b_that=-1.0
y = y_true + torch.randn(100) * 0.5 # thêm nhiễu

# ---- Khởi tạo tham số ----
w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

# ---- Vòng lặp huấn luyện ----
lr = 0.1 # Tăng lr lên một chút để hội tụ nhanh hơn trên Nitro 5
so_epoch = 200
lich_su = []

for epoch in range(so_epoch):
    # ??? Bước 1: Tính y_hat = w * X + b (Forward Pass)
    y_hat = w * X + b
    
    # ??? Bước 2: Tính MSE loss = mean((y_hat - y)^2)
    # Đây là hàm mục tiêu cần tối thiểu hóa
    loss = torch.mean((y_hat - y)**2)
    
    # ??? Bước 3: Tính gradient (Backward Pass)
    loss.backward()
    
    # ??? Bước 4: Cập nhật w, b (dùng torch.no_grad)
    # Chúng ta dùng in-place để cập nhật trực tiếp giá trị Tensor
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad
        
    # ??? Bước 5: Xoá gradient (Reset cho epoch tiếp theo)
    w.grad.zero_()
    b.grad.zero_()
    
    if epoch % 20 == 0:
        lich_su.append((epoch, loss.item(), w.item(), b.item()))
        print(f"Epoch {epoch:4d}: loss={loss.item():.4f}, "
              f"w={w.item():.4f}, b={b.item():.4f}")

print(f"\nKết quả cuối cùng: w={w.item():.4f} (Thật: 2.5), "
      f"b={b.item():.4f} (Thật: -1.0)")

# ---- Vẽ kết quả (Giữ nguyên phần vẽ của bạn) ----
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].scatter(X.numpy(), y.numpy(), s=10, alpha=0.5, label="Dữ liệu")
axes[0].plot(X.numpy(), (w * X + b).detach().numpy(), "r-", lw=2, label=f"Dự đoán (w={w.item():.2f})")
axes[0].plot(X.numpy(), y_true.numpy(), "g--", lw=1.5, label="Đường thật")
axes[0].legend(); axes[0].set_title("Hồi quy tuyến tính")

epochs_plot = [e for e, *_ in lich_su]
losses_plot = [l for _, l, *_ in lich_su]
axes[1].plot(epochs_plot, losses_plot, "b-o", markersize=4)
axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("MSE Loss")
axes[1].set_title("Loss curve"); axes[1].grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
